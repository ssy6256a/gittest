#### 인스턴스(멤버) 변수
- 객체마다 따로따로 가지는 데이터
- `self.변수명` 형태로 작성

```python
# 자동차 클래스 설계
# 속성 : 색상, 속도 => 멤버 변수
# 동작 : 전진, 후진, 속도올리기, 속도내리기.... => 멤버 메소드

class Car:
    def __init__(self,color,speed):
        # 인스턴스(멤버) 변수 -> self.변수
        self.color = color
        self.speed = speed


    # 인스턴스(멤버) 메소드(함수)
    def UpSpeed(self, value):
        self.speed += value



    def DownSpeed(self, value):
        self.speed -= value

    # 0x0000026D85A01A90 => 보고싶은 문자로 변경
    def __str__(self):
        return f"Car(color = {self.color}, speed = {self.speed})"
```

#### 클래스 메소드
- 클래스 자체를 다루는 메소드
- 첫번째 매개변수로 `self.` 대신  `cls`를 받음
- 클래스 변수 관리

```python
class SelfTest:
    @classmethod   # @ = 데코레이터

    # 클래스 메서드
    def function1(cls):
        print("function1 called")

    # 인스턴스(멤버) 메서드
    def function2(self):
        print("function2 called")

test1 = SelfTest()
# test1.function1()
SelfTest.function1() #클래스이름.클래스메소드
test1.function2()
```

```python
# 클래스 메소드 추가
class Dog:

    # 클래스 변수
    dog_count = 0
    species = "강아지"

    def __init__(self,name,age):
        self.name = name
        self.age  = age
        Dog.dog_count += 1

    def bark(self):
        return f"{self.name} : 멍멍"

    @classmethod
    def get_count(cls):
        return f"지금까지 만들어진 {cls.species}는 {cls.count}입니다."


dog1 = Dog("뽀삐",6)
dog2 = Dog("백구",7)

print("이름:",dog1.name,", 나이:", dog1.age,"살")
print("이름:",dog2.name,", 나이:", dog2.age,"살")

print(dog1.bark())
print(dog2.bark())

print(f"강아지는 총 {Dog.dog_count}마리 있다.")

```

- 객체의 상태나 클래스 변수를 전혀 건드리지 않을 때 사용

```python
class Car:

    # 클래스 변수
    car_count = 0

    def __init__(self,color,speed):
        # 인스턴스(멤버) 변수 -> self.변수
        self.color = color
        self.speed = speed
        Car.car_count += 1    # 클래스 변수는 self.로 접근 X


    # 인스턴스(멤버) 메소드(함수)
    def UpSpeed(self, value):
        self.speed += value



    def DownSpeed(self, value):
        self.speed -= value


    def __str__(self):
        return f"Car(color = {self.color}, speed = {self.speed})"
    
    @staticmethod
    def hello():
        print("안녕하세요")
    

car1 = Car("red",80)    
car2 = Car("white",70)    
car3 = Car("black",85)    

print(f"생산된 자동차 숫자는 {Car.car_count}대")

Car.hello()
```

| 구 분 | 첫 인자 | 접근 가능 | 주 용도 |
|----|----|----|----|
|`인스턴스 메서드`|`self`|인스턴스 변수, 클래스 변수 모두 | 객체별 동작|
|`@classmethod`|`cls`|클래스 변수만 | 클래스 단위 동작|
|`@staticmethod`|없음|없음(매개변수로 받은 것만) | 클래스와 관련은 있지만 독립적인 계산 |

```python
class Student:
    # 클래스 변수 - 공통적으로 사용할 변수 (전에 만들었던 카드 클래스 : 각각의 카드의 넓이와 높이는 모두 같음)
    school_name = "솔데스크"
    student_count = 0

    # 인스턴스(멤버) 변수
    def __init__(self,name):
        self.name = name
        # 전체 학생 수 
        Student.student_count += 1

    # 인스턴스(멤버) 메소드
    def show_info(self):
        print("=== 인스턴스 메서드 ===")

        # 인스턴스 변수에 접근 가능
        print(f"학생 이름 : {self.name}")

        # 클래스 변수에도 접근 가능
        print(f"학교 명 : {Student.school_name}")
        print(f"학생 수 : {Student.student_count}")


    @classmethod
    def show_school_info(cls):
        print("=== 클래스 메서드 ===")
        # print(f"학생 이름 : {self.name}") - 인스턴스 변수 접근(X)
        print(f"학교 명 : {Student.school_name}")  # - 클래스 변수 접근 가능
        print(f"학생 수 : {Student.student_count}") # - 클래스 변수 접근 가능
        

    @staticmethod
    def add_score(score1,score2):
        print("=== 정적 메서드 ===")
        # print(f"학생 이름 : {self.name}") - 인스턴스 변수 접근
        # print(f"학교 명 : {Student.school_name}")  # - 클래스 변수 접근 
        # print(f"학생 수 : {Student.student_count}") # - 클래스 변수 접근 (에러는 나지 않지만 용도가 아님)

        print(f"총점 : {score1 + score2}")


stu1 = Student("홍길동")
stu2 = Student("성춘향")

# 인스턴스 메서드 호출
stu1.show_info()
stu2.show_info()

# 클래스 메서드 호출
Student.show_school_info()

# 정적 메서드
Student.add_score(80,95)
```

```python
# tms
# Account 클래스 작성
# 멤버 변수 : 계좌주(owner), 잔액(balance)
#     메소드 : deposit(money) : 잔액 = 잔액 + 입금액
#     메소드 : withdraw(money) : 잔액 = 잔액 - 출금액
#     메소드 : showBalance() : 현재 잔액 출력

class Account:
    def __init__(self,owner,balance):
        self.owner = owner
        self.balance = balance

    def deposite(self,money):
        self.balance = self.balance + money
        return self.balance
    def withdraw(self,money):
        self.balance = self.balance - money
        return self.balance
    
    def showBalance(self):
            return (f"현재 {self.owner}님의 통장안의 잔액은 {self.balance}원 입니다.")

    

    
```

```python
# tms
# Order 클래스 작성
# 클래스 변수 : total_order
# 멤버 변수 : 메뉴(menu) , 가격(price), 수량(count)
# 메소드 : getTotalPrice() = 가격 * 수량 : 총 주문 금액 반환
#         showOrder() : 출력문 작성
#         메뉴 : 아메리카노
#         가격 : 4000원
#         수량 : 2개
#         총 주문 금액 : 8000원

class Order:
    total_order = 0
    def __init__(self, menu, price, count):
        self.menu = menu
        self.price = price
        self.count = count

    def getTotalPrice(self):
        Order.total_order = self.price * self.count
        return Order.total_order
    
    def showOrder(self):
        print(f"메뉴 : {self.menu}")
        print(f"가격 : {self.price}")
        print(f"수량 : {Order.total_order}")
        print(f"총 주문 금액 : {self.getTotalPrice}")

    def __str__(self):
        print(f"메뉴 : {self.menu}")
        print(f"가격 : {self.price}")
        print(f"수량 : {Order.total_order}")
        print(f"총 주문 금액 : {self.getTotalPrice}")

```

```python
client1 = Order("아이스 아메리카노", 4000, 2)
client1.showOrder()
```

#### 상속
- 기존 클래스(부모/상위)의 멤버 변수와 메서드를 새로운 (자식/하위)가 그대로 물려받는 것
- 코드 재사용성, 확장

```python
class Parent:
    def __init__(self):
        self.value = "테스트"
        print("Parent 클래스의 __init__() 호출")

    def test(self):
        print("Parent 클래스의 test() 호출")



class Child(Parent):
    pass

child = Child()
child.test()
print(child.value)
```

```python
# super() : 상위 클래스
class Child(Parent):
    def __init__(self):
        super().__init__() # <--
        print("Child 클래스의 __init__ 호출")

child = Child()
child.test()
print(child.value)        
```

```python
class Animal:
    def __init__(self,name):
        self.name = name

    def eat(self):
        print(f"{self.name}가 먹이를 먹는다.")

    def sound(self):
        print(f"{self.name}가 소리를 낸다.")

class Dog(Animal):
    def __init__(self, name, kind):
        super().__init__(name)
        self.kind = kind

    def fetch(self):
        print(f"{self.name}가 공을 물어옵니다.")


dog = Dog("뽀삐","쿠키")
dog.eat()
dog.sound()
dog.fetch()
```

```python
# Audio : 멤버변수 - power, volume / 메소드 - 전원 조절, 볼륨 조절 + tune()
# TV : 멤버변수 - power, volume, size / 메소드 - 전원 조절, 볼륨 조절 + watch()
# 먼저 audio와 tv의 중복되는 것으로 부모 클래스를 만들고 audio와 tv의 고유적인 메소드는 개별 클래스로 넣음

class AudioVisual:
    def __init__(self,power,volume):
        self.power = power
        self.volume = volume

    def switch(self, on_off):
        self.power = on_off

    def setvolume(self, volume):
        self.volume = volume


class Audio(AudioVisual):
    def __init__(self, power, volume):
        super().__init__(power, volume)

    def tune(self):
        print("La la la...." if self.power else "turn it on")

class TV(AudioVisual):
    def __init__(self, power, volume, size):
        super().__init__(power, volume)
        self.size = size

    def watch(self):
        print("have fun!" if self.power else "switch on")
        

tv = TV(False, 12, 50)
# tv켜기
tv.switch(True)
# tv 보기
tv.watch()

audio = Audio(True, 15)
audio.setvolume(10)
audio.tune()
audio.switch(False)
```

#### 오버라이딩
- 부모 클래스에 있는 메서드를 자식 클래스에서 같은 이름으로 다시 정의 해서 동작 내용 변경 (자식 이기는 부모 없다. ; 자식은 부모와 다른 경험을 한다.)


```python
class Animal:
    def __init__(self, name):
        self.name = name

    def sound(self):
        print(f"{self.name}가 울음 소리를 냅니다.")

class Dog(Animal):
    def sound(self):
        # return super().sound()
        super().sound()  ## 오버라이딩 시 동일한 부모 메소드를 호출하고 싶다면 super().sound() 로 호출 가능
        print(f"{self.name} : 멍멍!!")

class Cat(Animal):
    def sound(self):
        # return super().sound()
        super().sound()
        print(f"{self.name} : 야옹!!")

dog = Dog("뽀삐")
cat = Cat("나비")

dog.sound()
cat.sound()
```

```python
# tms
# Employee 부모클래스 :
# 멤버 변수 : 이름(name), 급여(pay) 
#     메소드 : showInfo() => 이름, 급여 출력

# Manager 자식 클래스 :
# 추가 멤버 변수 : 보너스(bonus)
#     메소드 : getTotalPay() : 급여 + 보너스 출력

class Employee:
    def __init__(self,name,pay):
        self.name = name
        self.pay = pay

    def showInfo(self):
        print(f"이름 : {self.name}\n급여 : {self.pay}")

# em = Employee("hong", 10000)
# em.showInfo()

class Manager(Employee):
    def __init__(self, name, pay,bonus):
        self.bonus = bonus
        super().__init__(name, pay)

    def getTotalPay(self):
        print(f"총급여 : {self.pay + self.bonus}원")
    
employee1 = Manager("홍길동",10000, 3000 )
employee2 = Employee("성춘향",10000)

employee1.showInfo()
employee1.getTotalPay()

employee2.showInfo()

```

```python
# tms
# Shape 부모 클래스
# 메서드 : area() => "도형의 넓이" 출력

# Rectangle 자식 클래스
# 멤버 변수 : width , height
# 메서드 : area() => 직사각형의 넓이 출력

# Circle 자식 클래스
# 멤버 변수 : 반지름(radius)
# 메서드 : area() => 원의 넓이 출력

class Shape():
    def __init__(self):
        pass
    def area(self):
        print("도형의 넓이")

a = Shape()
a.area()

class Rectangle(Shape):
    def __init__(self,width,height):
        self.width = width
        self.height = height
        super().__init__()

    def area(self):
        print(f"사격형의 넓이 : {self.width * self.height}")
    
b = Rectangle(10,12)
b.area()
    
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
        super().__init__()

    def area(self):
        print(f"원의 넓이 : {self.radius ** 2 * 3.14}")

c = Circle(5)
c.area()
```
