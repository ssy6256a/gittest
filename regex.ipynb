#### 정규표현식(정규식)
- 복잡한 문자열을 처리할 때 사용하는 방법
- 파이썬만의 고유한 문법이 아니라 문자열을 처리하는 모든 곳에서 사용하는 일종의 형식 언어

# 패턴을 특정 문자열이 포함하고 있는가?
| 메서드 | 목적 |
|-----|----|
| match | 문자열의 처음부터 정규식과 매치되는지 조사 |
| search | 문자열의 전체를 검색하여 정규식과 매치되는지 조사 |
| findall | 정규식과 매치되는 모든 문자열(substring)을 리스트로 리턴 |
| fimditer | 정규식과 매치되는 모든 문자열(substring)을 이터레이터 객체로 리턴 |


```python
# [문자 나열] : [] 안에 들어있는 문자들과 매치하느냐? :: match() 
p = re.compile("[abc]")

print(p.match("a"))
print(p.match("before"))
print(p.match("dude"))
```

```python
# print(p.match("a")) 를 찍으면 <re.Match object; span=(0, 1), match='a'> 이렇게 나오는데 내가 원하는 자료만 보고 싶으면
p = re.compile("[abc]")
m = (p.match("a"))

print(f"패턴 시작 : {m.start()}")
print(f"패턴 끝 : {m.end()}")
print(f"패턴과 일치하는 문자열 반환 : {m.group()}")
print(f"일치의 시작,끝 반환 : {m.span()}")

if m:
    print(f"match found {m.group()}")
else:
    print("no match")
```

- 자주 사용하는 문자 표현식

| 정규표현식 | 설명 |
|-----|-----|
|\d|숫자와 매치 [0-9]와 동일한 표현식|
|\D|숫자가 아닌 것과 매치 [^0-9]와 동일한 표현식|
|\s|화이트스페이스(스페이스,탭,줄바꿈 등)문자와 매치 [빈칸\t\n\r\f\v]와 동일한 표현식|
|\S|화이트스페이스(스페이스,탭,줄바꿈 등)문자가 아닌 것과 매치 [빈칸\t\n\r\f\v]와 동일한 표현식|
|\w|문자 + 숫자와 매치 [a-zA-Z0-9]와 동일한 표현식|
|\W|문자 + 숫자와 매치 [^a-zA-Z0-9]와 동일한 표현식|




```python
#원본 문자열에서 0~9 인 문자 찾기
p = re.compile("[0-9]+")
p = re.compile(r"\d+")
print(p.match("012 abc"))

#원본 문자열에서 0~9 가 아닌 문자 찾기
# p = re.compile("[^0-9]+")
p = re.compile(r"\D+")
print(p.match("012 abc"))

#원본 문자열에서 공백,탭,엔터 인 문자 찾기
# p = re.compile("[ \t\n\r\f\v]+")
p = re.compile(r"\s+")
print(p.match("012 abc"))

#원본 문자열에서 공백,탭,엔터 가 아닌 문자 찾기
# p = re.compile("[^ \t\n\r\f\v]+")
p = re.compile(r"\S+")
print(p.match("012 abc"))

#원본 문자열에서 대소문자,숫자,_를 포함한 문자 찾기
# p = re.compile("[a-zA-Z0-9]+")
p = re.compile(r"\w+")
print(p.match("012 abc"))

#원본 문자열에서 대소문자,숫자,_를 포함하지 않은 문자 찾기
# p = re.compile("[^a-zA-Z0-9]+")
p = re.compile(r"\W+")
print(p.match("012 abc"))
```

```python
# re.I : ignorecase

# 영어가 포함되어 있는지?

p = re.compile("[a-z]+", re.I) # == p = re.compile("[a-zA-z]+")
print(p.match("python"))
print(p.match("Python"))
print(p.match("PYTHON"))
```

```python
# re.M : MULTILINE 여러줄
data = """python one
life if too short
python two
you need python
python three
"""
# 대괄호에 포함되어 있지 않은 ^ 의 의미 : 시작하는
p = re.compile(r"^python\s\w+") # == python 으로 시작하고/ 스페이스 문자 하나/ 영어,숫자,_인 문자 찾아줘
print(p.findall(data)) # --> ['python one']

p = re.compile(r"^python\s\w+", re.M)
print(p.findall(data))
```

```python
# () : 그루핑(grouping)할 때 사용 // ()를 치면 그룹이 지어져서 1,2...으로 명령할 수 있음

#ABCABC 문자열이 반복되는지 조사

p = re.compile("[ABC]+")
m = p.search("ABCABCABC OK?")
print(m)
print(m.group())
# print(m.group(1)) : IndexError: no such group
```

```python
#TMS
# 아이디 비밀번호 전화번호
# 패턴에 맞을때까지 입력 받기
n=0
# 아이디 검사 조건
# Qwer!12
# 
# 01012345678

# 영문 대문자로 시작 ^[A-Z], 영문,숫자,특수문자(!@#$%) 포함, 6~15자
while n <=3:
    p = re.compile(r"^[A-Z][a-zA-Z0-9!@#$%]{5,14}") #$
    num = input("아이디를 입력해주세요")
    if re.fullmatch(p, num):
        print("확인되었습니다.")
        n += 1
    else:
        print("다시 입력해주세요")
# 비밀번호
# 비밀번호 검사
# 영문 대소문자, 숫자, 특수문자(~~) 포함, 8자 이상
    p = re.compile(r"^[A-Z][a-zA-Z0-9!@#$%]{8,}")
    password = input("비밀번호를 입력해주세요")
    if re.fullmatch(p, password):
        print("확인되었습니다.")
        n += 1
    else:
        print("다시 입력해주세요")

# 전화번호 010-1234-5678 / 01012345678
    p = re.compile(r"^010-?(\d{4})-?(\d{4})")
    call = input("전화번호를 입력해주세요")
    if re.fullmatch(p, call):
        print("확인되었습니다.")
        n += 1
    else:
        print("다시 입력해주세요")

print(f"입력하신 아이디:{num}\n입력하신 비밀번호:{password}\n입력하신 전화번호{call}")

# fullmatch()
```

```python

# id = re.compile("[A-Z][a-z0-9!@#%&]{5,14}") ==> A12345678 이것도 가능

# 대문자로 시작, 영어소문자 하나 이상, 숫자 하나이상, 특수문자 하나이상

# 정규식의 기본형태 : 패턴을 소비(~= 체크리스트처럼)한다 => 기억하지 않음
id = re.compile(r"^(?=.*[A-Z])(?=.*[a-z])(?=.*\d)(?=.*[!@#%&])[A-Z][a-z0-9!@#$%&]{5,14}$")

print(id.fullmatch("A10101010"))
print(id.fullmatch("abc@123k"))
print(id.fullmatch("A1bc01$&0"))
```

```python
# 영문 대소문자, 숫자, 특수문자(~~) 포함, 8자 이상
pwd = re.compile(r"^(?=.*[A-Z])(?=.*[a-z])(?=.*\d)(?=.*[!@#%&])[A-Za-z0-9!@#$%&]{8,}$")

print(pwd.fullmatch("A10101010"))
print(pwd.fullmatch("abc@123k"))
print(pwd.fullmatch("A1bc01$&0"))
```

```python
# greedy 와, non-greedy

s = "<html><head><title>Title</title></head></html>"

#greedy
p = re.compile("<.+>")
print(p.search(s).group()) 

#non-greedy
# *?, {m,n}?
p = re.compile("<.+?>")
print(p.search(s).group()) 
```
