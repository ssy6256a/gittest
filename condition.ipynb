````python
# 주차장 프로그램
# 사용자에게 메뉴 제공
# [1] 주차 | [2] 출차 | [3] 종료

# 사용자는 메뉴 선택 
# 1번 선택 시 
# 자동차 A ~ E
# A 자동차 입차. 주차장 상태 ==> ['A']
# 1
# B 자동차 입차. 주차장 상태 ==> ['B']
# 2
# B 자동차 출차. 주차장 상태 ==> ['A']
# E 자동차 입차. 주차장 상태 ==> ['A', 'B', 'C', 'D', 'E']
# 1
# 주차장 만차

parking_lot = []
top, car_name = 0, 'A'


while True:
    no = int(input("[1] 주차 | [2] 출차 | [3] 종료"))

    if no <= 3:
        if no ==1:
            # top < 5 주차장 입차. 주차장 상태 ==> ['A']
            # A 자동차 입차. 주차장 상태 ==> ['A']
            # top >= 5 주차장 만차
            if top >= 5:
                print("주차장 만차")
            else:
                parking_lot.append(car_name)
                print(f"{car_name} 자동차 입차. 주차장 상태 ==> {parking_lot}")
                car_name = chr(ord(car_name)+1)
                top = top + 1
        elif no == 2:
            # if top > 0 출차
            # B 자동차 출차. 주차장 상태 ==> ['A']
            # top -1, car_name -1
            # 빠져나갈 자동차 없음
            if top > 0:
                out_car = parking_lot.pop()
                print(f"{out_car} 자동차 출차. 주차장 상태 ==> {parking_lot}")
                car_name = chr(ord(car_name)-1)
                top = top - 1
            else:
                print("주차장 내부에 자동차가 없습니다.")
            
        else:
            print("프로그램 종료")
            break
    else:
        print("번호를 확인해 주세요")
````

````python
# 새로운 리스트 생성 : 리스트 컴프리헨션
# 리스트 안에 for 문을 이용해서 그 조건에 맞는 새로운 리스트 생성

print([i for i in range(1,7)])
print([i * i for i in range(5)])
print([i * 2 for i in [1,2,3,4,5]])
print([[i, i+1, i+2] for i in [1,2,3]])

# 1 ~ 10 사이의 숫자 중에서 홀수 리스트 생성
odd_list = []
for i in range(1,11):
    if i % 2 == 1:
        odd_list.append(i)

print(odd_list)

odd_list2 = [i for i in range(1,11) if i % 2 == 1]
print(odd_list2)
````

````python
q_list = ["갑","을", "병", "정", "경"]

# q_list에서 정을 제외한 요소를 갖는 새로운 리스트 생성
# for

newlist = []
for i in q_list:
    if i !='정':
        newlist.append(i)

print(newlist)

newlist1 = []
for i in q_list:
    if i == "정": continue
    newlist1.append(i)
    
print(newlist1)





# 리스트 컴프리헨션
q_list2 = ["갑","을", "병", "정", "경"]
q_list2 = [i for i in q_list2 if i != "정"]
q_list2

````

````python
# 왜 '경'만 나오지;
q_list = ["갑","을", "병", "정", "경"]

newlist1 = []
for i in q_list:
    if i == "정": continue
newlist1.append(i)
    
print(newlist1)
````

````python
q1 = ['nice','study','python','anaconda','!']
# 5글자 이상의 단어만 새로운 리스트로 생성
# len()


q1 = [i for i in q1 if len(i) >= 5]
print(q1)
````

````python
q2 = ['A','b','c','D','e','F','G','h']
# 소문자만 추출 후 새로운 리스트로 생성

newlist = []
newlist = [i for i in q2 if i.islower()]
newlist2 = [i for i in q2 if i.isupper()]

print(newlist)
print(newlist2)

# "X".islower()
# "X".isupper()
````
